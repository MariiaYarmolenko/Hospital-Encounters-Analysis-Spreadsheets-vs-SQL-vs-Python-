```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Data/Hospital+Patient+Records/encounters.csv')
```

```python
print(df.head())
print(df.info())
```

```python
df['START'] = pd.to_datetime(df['START'])
df['STOP'] = pd.to_datetime(df['STOP'])

print(df[['START', 'STOP']].dtypes)
```

```python
df['duration_hours'] = (df['STOP'] - df['START']).dt.total_seconds() / 3600

print(df[['START', 'STOP', 'duration_hours']].head())
```

```python
analysis_pd = df.groupby('ENCOUNTERCLASS').agg({
    'TOTAL_CLAIM_COST': ['count', 'sum', 'mean'],
    'duration_hours': 'mean'
})

analysis_pd.columns = ['total_visits', 'total_cost', 'avg_cost', 'avg_duration_hours']
print(analysis_pd.round(2))
```
